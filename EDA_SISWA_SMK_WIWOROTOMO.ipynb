### 1. IMPORT DATA

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", 100)
pd.set_option("display.max_colwidth", 60)

OUTPUT_DIR = Path("outputs")
OUTPUT_DIR.mkdir(exist_ok=True)
```

### 2. MEMUAT DATASET

```python
import pandas as pd

# Lokasi file CSV di Google Drive
file_path = '/content/drive/MyDrive/Proyek/S2/DATASCIENCE/DATA CALON SISWA SMK WIWOROTOMO 2026_2027.csv'

# Membaca CSV dengan encoding alternatif
data = pd.read_csv(file_path, encoding='latin1')

# Menampilkan ukuran dataset
print("Jumlah baris dan kolom:", data.shape)

# Menampilkan 5 data pertama
display(data.head())
```

### 3. PEMERIKSAAN AWAL


```python
print("Dimensi dataset:", data.shape)
print("\\nDaftar kolom:")
for i, col in enumerate(data.columns, start=1):
    print(f"{i:02d}. {col}")

print("\\nInformasi dataset:")
data.info()

print("\\nStatistik deskriptif:")
display(data.describe(include="all").T)

print("\\nJumlah baris duplikat:", data.duplicated().sum())
```

### 4. PEMERIKSAAN KOLOM UNNAMED

```python
unnamed_cols = [c for c in data.columns if str(c).startswith("Unnamed")]
if unnamed_cols:
    unnamed_summary = pd.DataFrame({
        "jumlah_terisi": data[unnamed_cols].notna().sum(),
        "jumlah_unik": data[unnamed_cols].nunique(dropna=True)
    })
    display(unnamed_summary)

    # Tampilkan contoh nilai non-kosong tanpa menampilkan identitas individu
    for col in unnamed_cols:
        print(f"{col}: {data[col].notna().sum()} nilai terisi")
else:
    print("Tidak ditemukan kolom Unnamed.")
```

### 5. PEMBERSIHAN DATA UTAMA

```python
# Salin data mentah agar tidak berubah
df = data.copy()
df.columns = df.columns.astype(str).str.strip()

# Kolom identitas dan informasi pribadi yang tidak diperlukan untuk EDA agregat
kolom_privasi = [
    "NO. Pendaftaran",
    "Nama Peserta Didik (CAPSLOCK)",
    "NISN",
    "Tanggal Lahir",
    "ALAMAT PESERTA DIDIK (Desa, RT/RW, Kec. Kab. Provinsi)",
    "Kode Pos",
    "Nama Ayah (CAPSLOCK)",
    "Nama Ibu  (CAPSLOCK)",
    "Nama Wali",
    "Alamat Wali",
    "Nomor HP Wali",
    "Rekomendasi",
    "Keterangan / Lain-lain"
]

# Kolom Unnamed disisihkan dari analisis utama, tetapi tetap ada di data mentah
kolom_unnamed = [c for c in df.columns if str(c).startswith("Unnamed")]

# Data analisis: hanya kolom utama yang bukan identitas langsung
df_eda = df.drop(
    columns=[c for c in kolom_privasi + kolom_unnamed if c in df.columns],
    errors="ignore"
).copy()

# Rapikan spasi pada kolom teks dan ubah string kosong menjadi NaN
for col in df_eda.select_dtypes(include="object").columns:
    df_eda[col] = df_eda[col].str.strip()
    df_eda[col] = df_eda[col].replace({"": np.nan, "-": np.nan, "—": np.nan})

print("Ukuran data mentah:", df.shape)
print("Ukuran data untuk EDA:", df_eda.shape)
print("Duplikasi pada data utama:", df_eda.duplicated().sum())
display(df_eda.head())
```

### 6. ANALISIS MISSING VALUES

```python
missing = df_eda.isna().sum().to_frame("jumlah_kosong")
missing["persentase_kosong"] = (missing["jumlah_kosong"] / len(df_eda) * 100).round(2)
missing = missing.sort_values("persentase_kosong", ascending=False)
display(missing)

plt.figure(figsize=(11, 6))
plot_missing = missing[missing["jumlah_kosong"] > 0]
sns.barplot(data=plot_missing.reset_index(), x="persentase_kosong", y="index")
plt.title("Persentase Missing Value per Kolom")
plt.xlabel("Missing value (%)")
plt.ylabel("Kolom")
plt.tight_layout()
plt.show()
```

### 7. FUNGSI ANALISIS KATEGORI

```python
def plot_kategori(dataframe, kolom, top_n=20, judul=None):
    if kolom not in dataframe.columns:
        print(f"Kolom tidak ditemukan: {kolom}")
        return
    frekuensi = dataframe[kolom].dropna().astype(str).value_counts().head(top_n)
    if frekuensi.empty:
        print(f"Tidak ada data pada kolom: {kolom}")
        return

    plt.figure(figsize=(10, max(4, len(frekuensi) * 0.38)))
    sns.barplot(x=frekuensi.values, y=frekuensi.index)
    plt.title(judul or f"Distribusi {kolom}")
    plt.xlabel("Jumlah calon siswa")
    plt.ylabel(kolom)
    plt.tight_layout()
    plt.show()
    display(frekuensi.rename("jumlah").to_frame())
```

### 8. ANALISIS PENDAFTAR BERDASARKAN JURUSAN

```python
plot_kategori(df_eda, "Jurusan", top_n=30,
              judul="Jumlah Calon Siswa Berdasarkan Jurusan")
```

### 9. ANALISIS IMBALANCED DATA ANTAR JURUSAN

```python
import pandas as pd
import matplotlib.pyplot as plt

kolom_jurusan = 'Jurusan'

# Menghitung jumlah pendaftar
jumlah_jurusan = df[kolom_jurusan].value_counts()

# Persentase
persentase_jurusan = df[kolom_jurusan].value_counts(normalize=True) * 100

# Tabel distribusi
df_imbalance = pd.DataFrame({
    'Jumlah Pendaftar': jumlah_jurusan,
    'Persentase (%)': persentase_jurusan.round(2)
})

df_imbalance = df_imbalance.sort_values(
    by='Jumlah Pendaftar',
    ascending=False
)

display(df_imbalance)

# Rasio kelas terbesar terhadap terkecil
jumlah_max = jumlah_jurusan.max()
jumlah_min = jumlah_jurusan.min()

rasio = jumlah_max / jumlah_min

print("Jumlah kelas terbesar:", jumlah_max)
print("Jumlah kelas terkecil:", jumlah_min)
print(f"Rasio kelas terbesar : terkecil = {rasio:.2f} : 1")

# Visualisasi
plt.figure(figsize=(10, 5))
jumlah_jurusan.sort_values().plot(kind='barh')

plt.title('Distribusi Pendaftar Antarjurusan')
plt.xlabel('Jumlah Pendaftar')
plt.ylabel('Jurusan')
plt.tight_layout()
plt.show()
```

### 12. MENORMALISASI NAMA ASAL SEKOLAH

```python
import pandas as pd
import re

# Salin data agar data asli tidak berubah
df_normalisasi = df.copy()

kolom_sekolah = 'SEKOLAH ASAL (SMP/MTs)'

def normalisasi_sekolah(teks):
    if pd.isna(teks):
        return teks

    teks = str(teks).strip().upper()

    # Normalisasi spasi berlebih
    teks = re.sub(r'\s+', ' ', teks)

    # Menyamakan variasi penulisan sekolah negeri
    teks = re.sub(r'\bSMP\s*N\b', 'SMP NEGERI', teks)
    teks = re.sub(r'\bSMA\s*N\b', 'SMA NEGERI', teks)
    teks = re.sub(r'\bSMK\s*N\b', 'SMK NEGERI', teks)

    # Menyamakan singkatan negeri
    teks = re.sub(r'\bSMPN\b', 'SMP NEGERI', teks)
    teks = re.sub(r'\bSMAN\b', 'SMA NEGERI', teks)
    teks = re.sub(r'\bSMKN\b', 'SMK NEGERI', teks)

    # Merapikan spasi kembali
    teks = re.sub(r'\s+', ' ', teks).strip()

    return teks

# Membuat kolom hasil normalisasi
df_normalisasi['Asal Sekolah Normal'] = (
    df_normalisasi[kolom_sekolah].apply(normalisasi_sekolah)
)

# Perbandingan nama sebelum dan sesudah normalisasi
perbandingan_sekolah = df_normalisasi[
    [kolom_sekolah, 'Asal Sekolah Normal']
].drop_duplicates()

display(perbandingan_sekolah.head(30))
```

```python
plot_kategori(df_normalisasi, "Asal Sekolah Normal", top_n=20,
              judul="20 Sekolah Asal dengan Pendaftar Terbanyak")
```

```python
jumlah_sekolah_sebelum = df[kolom_sekolah].nunique()
jumlah_sekolah_sesudah = df_normalisasi[
    'Asal Sekolah Normal'
].nunique()

print("Jumlah kategori sekolah sebelum normalisasi:",
      jumlah_sekolah_sebelum)

print("Jumlah kategori sekolah sesudah normalisasi:",
      jumlah_sekolah_sesudah)

print("Kategori sekolah yang berhasil disederhanakan:",
      jumlah_sekolah_sebelum - jumlah_sekolah_sesudah)
```

```python
import matplotlib.pyplot as plt

top_sekolah_normal = (
    df_normalisasi['Asal Sekolah Normal']
    .value_counts()
    .head(20)
)

display(top_sekolah_normal)

plt.figure(figsize=(10, 7))

top_sekolah_normal.sort_values().plot(
    kind='barh'
)

plt.title('20 Sekolah Asal Terbanyak Setelah Normalisasi')
plt.xlabel('Jumlah Pendaftar')
plt.ylabel('Sekolah Asal')
plt.tight_layout()
plt.show()
```

### 13. ANALISIS DATA NUMERIK

```python
kolom_numerik = ["Tinggi Badan", "Berat Badan", "Anak Ke-", "Jumlah Saudara",
                 "Tahun Lahir Ayah", "Tahun Lahir Ibu"]

for col in kolom_numerik:
    if col in df_eda.columns:
        df_eda[col] = pd.to_numeric(
            df_eda[col].astype(str).str.replace(",", ".", regex=False),
            errors="coerce"
        )

kolom_tersedia = [c for c in kolom_numerik if c in df_eda.columns]
if kolom_tersedia:
    display(df_eda[kolom_tersedia].describe().T)
    for col in kolom_tersedia:
        nilai = df_eda[col].dropna()
        if not nilai.empty:
            plt.figure(figsize=(8, 4))
            sns.histplot(nilai, kde=True)
            plt.title(f"Distribusi {col}")
            plt.xlabel(col)
            plt.tight_layout()
            plt.show()
```

### 15. TABULASI SILANG JURUSAN DAN JENIS KELAMIN

```python
if {"Jurusan", "Jenis Kelamin"}.issubset(df_eda.columns):
    tabel = pd.crosstab(df_eda["Jurusan"], df_eda["Jenis Kelamin"], margins=True)
    display(tabel)

    tabel_plot = pd.crosstab(df_eda["Jurusan"], df_eda["Jenis Kelamin"])
    tabel_plot.plot(kind="bar", stacked=True, figsize=(11, 6))
    plt.title("Komposisi Jenis Kelamin pada Setiap Jurusan")
    plt.xlabel("Jurusan")
    plt.ylabel("Jumlah calon siswa")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```
